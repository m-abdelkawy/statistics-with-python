```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from pathlib import Path
```

```python
# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 3 * np.pi, 0.1)
y = np.sin(x)

# Plot the points using matplotlib
plt.plot(x, y)
```

```python
plt.grid(True)
plt.plot(x, y, linestyle="dotted", lw=5, color="orange")
plt.ylabel("Y", size=15)
plt.xlabel("X", size=15)
```

```python
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.grid(True)

plt.plot(x, y_sin)
plt.plot(x, y_cos)

plt.xlabel("x values", size=15)
plt.ylabel("y values", size=15)

plt.title("Sin and Cos")
plt.legend(["Sine", "Cosine"])
```

```python
plt.grid(True)
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y__cos = np.cos(x)

plt.subplot(2, 2, 1)
plt.plot(x, y_sin)
plt.title("Sine")

plt.subplot(2, 2, 2)
plt.plot(x, y_cos)
plt.title("Cosine")
```

```python
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

fig, axes = plt.subplots(2, 1)

axes[0].plot(x, y_sin)
axes[0].set_title("Sine")
axes[0].grid(True)

axes[1].plot(x, y_cos)
axes[1].set_title("Cosine")
axes[1].grid(True)
```

Seaborn:
Seaborn is essentially a high-level interface to Matplotlib that is intended to make statistical data visualization easier. Any plot that can be made in Seaborn can also be made in Matplotlib, but certain types of graphs are easier to make in Seaborn (and some others may be easier to make in Matplotlib).

```python
# Basic Scatter plot:

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"

plt.grid(True)
df = pd.read_csv(fname)

sns.scatterplot(x="Wingspan", y="CWDistance", hue="Gender", data=df)

# display(df)
```

```python
# Basic Swarm plot: when one var is categorical

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"

plt.grid(True)
df = pd.read_csv(fname)

sns.swarmplot(x="Gender", y="CWDistance", hue="Gender", data=df)
```

```python
# Basic Box plot:

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"

plt.grid(True)
df = pd.read_csv(fname)

sns.boxplot(data=df, y="Height")
```

```python
# Basic Box plot:

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"
df = pd.read_csv(fname)

fig, ax = plt.subplots(2, 1)
# ax.grid(True)

sns.boxplot(data=df, x="Gender", y="Height", hue="Gender", ax=ax[0])

sns.boxplot(data=df.loc[:, ["Age", "Height", "Wingspan", "CWDistance", "Score"]])
```

```python
# Basic Box plot only for females

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"
df = pd.read_csv(fname)

fig, ax = plt.subplots()

sns.boxplot(data=df.loc[df.Gender=="F", ["Age", "Height", "Wingspan"]], ax=ax)
```

```python
#Histogram:

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"
df = pd.read_csv(fname)

fig, ax = plt.subplots()

sns.histplot(data=df.loc[df.Gender=="F", "CWDistance"], ax=ax)
sns.displot(data=df.loc[df.Gender=="F", "CWDistance"]) # figure level function
```

```python
# count plot:

project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "cartwheeldata.csv"
df = pd.read_csv(fname)

sns.countplot(data=df, x="Gender")
plt.xticks(rotation=-45)
```
