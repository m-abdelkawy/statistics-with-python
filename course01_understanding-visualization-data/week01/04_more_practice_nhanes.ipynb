```python
project_root = Path.cwd().parents[0]
fname = project_root / "datasets" / "nhanes_2015_2016.csv"

df = pd.read_csv(fname)
```

#### Question 1
Relabel the marital status variable DMDMARTL to have brief but informative character labels. Then construct a frequency table of these values for all people, then for women only and for men only. Then construct these three frequency tables using only people whose age is between 30 and 40.

```python
ms_label = {1: "Married", 2: "Widowed", 3: "Divorced", 4: "Separated", 5: "NeverMarried", 6: "Cohabitating", 77: "Refused", 99: "Unknown"}

df["DMDMARTLx"] = df["DMDMARTL"].replace(ms_label)

marital_status = df["DMDMARTLx"].value_counts()
percentage = marital_status / marital_status.sum()

res = pd.DataFrame({
    "count":marital_status,
    "percentage": percentage
})

print("Marital status stats for all people: ")
display(res.style.format({"percentage":"{:.2%}"}))

```

```python
## replace gender with a label:
gender_label = {1: "Male", 2: "Female"}
df["RIAGENDRx"] = df["RIAGENDR"].replace(gender_label)

# groupby() is one of the most important pandas operations for split → apply → combine analysis
# it returns a pandas object of type "DataFrameGroupBy"
grouped = df.groupby("RIAGENDRx")
# grouped.groups
for name, group in grouped:
    print(f"marital status stats for : {name}s")
    x = group["DMDMARTLx"].value_counts()
    res = pd.DataFrame({
        "count": x,
        "percentage": x/x.sum()
    })
    display(res.style.format({"percentage": "{:.2%}"}))
```

```python
print("#### people between 30 and 40 ####")
df3040 = df.query('RIDAGEYR >= 30 & RIDAGEYR <= 40')
grouped = df3040.groupby("RIAGENDRx")
for name, group in grouped:
    print(f"marital status stats for : {name}s")
    x = group["DMDMARTLx"].value_counts()
    res = pd.DataFrame({
        "count": x,
        "percentage": x/x.sum()
    })
    display(res.style.format({"percentage": "{:.2%}"}))
```
